```python
import requests

URL = "https://rehab-unpopular-recast.ngrok-free.dev/summarize"
headers = {"Authorization": "Bearer secret123"}
payload = {"url": "https://www.youtube.com/watch?v=o-wv_szZ0V0"}

res = requests.post(URL, headers=headers, json=payload)
result = res.json()

print("===== Final Summary =====")
print(result["summary"])
print("\n===== Sections =====")
for i, p in enumerate(result["sections"], 1):
    print(f"{i}. {p}")
```
